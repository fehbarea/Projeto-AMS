def decidir(registro: dict, estado: EstadoDaFrota) -> str:

    if registro["vibracao_mm_s"] > LIMITE_VIBRACAO or \
            registro["temperatura_c"] > LIMITE_TEMPERATURA:
        return "parar_para_manutencao"
    if estado.falhas_recentes(registro["tipo_maquina"]) >= FALHAS_PARA_INSPECAO:
        return "inspecionar_antes_de_operar"
    return "seguir_operando"

**segunda linha de base da D2:**
    
- Foram usados os mesmo limiares para os sensores do baseline anterior, porém, foi adicionado que quando os sensores tiverem perto do limiar e já tiver ocorrido 1 falha ou muito tempo desde a ultima manutenção, deve ser inspecionado.

def decidir_b(registro, estado):
    vibracao = registro["vibracao_mm_s"]
    temepratura = registro["temperatura_c"]
    sensores_alterados = False
    recomendacao = "seguir_operando"
    limite_vibracao = 7.0  
    limite_temperatura = 78.0    

    if vibracao >= limite_vibracao or \
            temepratura >= limite_temperatura:
        return "parar_para_manutencao"
    
    if vibracao >= (limite_vibracao * 0.95) or \
            temepratura >= (limite_temperatura * 0.95):
        sensores_alterados = True

    falhas = estado.falhas_recentes(registro["tipo_maquina"])
    if falhas >= 2 :
        recomendacao = "inspecionar_antes_de_operar" 
    elif sensores_alterados and registro["horas_desde_manutencao"] >= 150:
        recomendacao = "inspecionar_antes_de_operar"
    elif falhas == 1 and sensores_alterados: 
        recomendacao = "inspecionar_antes_de_operar"

    return recomendacao
